### 4. Calculate sate-wise :
### A. Total Number of Population.
### B. Total Number of Population with different religions.

```python
df.groupby('State_name')['Population'].sum()
```

```python
df.columns
```

```python
data = df.groupby('State_name')[['Hindus', 'Muslims', 'Christians', 'Sikhs', 'Buddhists', 'Jains']]
for i in data:
    print(i , "\n")
```

```python
df.groupby('State_name')[['Hindus', 'Muslims', 'Christians', 'Sikhs', 'Buddhists', 'Jains']].sum().sort_values(by='Hindus' , ascending=False)
```
