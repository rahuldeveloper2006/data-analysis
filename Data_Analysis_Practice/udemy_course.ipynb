```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df.drop_duplicates(inplace=True)
```

```python
# convert price column from string to integer
df['price'] = df['price'].replace({"Free" : 0})
```

```python
df['price'] = df['price'].astype(np.int32)
```

```python
df.info()
```

### 1. What are all different subjects for which Udemy is offering courses ?

```python
df['subject'].unique()
```

### 2. Which subject has the maximum number of courses ?

```python
df1 = df['subject'].value_counts()
max = df['subject'].value_counts().values.max()
print(f"{df1[df1[df1.index] == max]} has the maximum number of Courses")
```

```python
df['subject'].value_counts()
```

### 3. show all the courses which are free of cost.

```python
df[['course_title' , 'subject' , 'price']][df['price'] == 0]
```

### 4. Show all the courses which are paid . 

```python
df[['course_title' , 'subject' , 'price']][df['price'] != 0]
```

### 5. Which are Top selling Courses ?

```python
df[['course_title' , 'subject' , 'price' , 'num_subscribers']][df['num_subscribers'] == df['num_subscribers'].max()]
```

```python
df.sort_values( by='num_subscribers' , ascending=False).head(10)
```

### 6. which are Least selling courses ?

```python
df[['course_title' , 'subject' , 'price' , 'num_subscribers']][df['num_subscribers'] == df['num_subscribers'].min()]
```

### 7. show all courses of graphic design where the price is below 100 ?

```python
df[(df['subject'] == 'Graphic Design') & (df['price'] < 100)]
```

### 8. List out all the courses that are related with 'Python'.

```python
df[df['course_title'].str.contains('Python')]
```

### 9. What are courses that published in year 2015 ?

```python
df['published_timestamp'].info()
```

```python
# now we convert from string to date-time data type
df['published_timestamp'] = pd.to_datetime(df['published_timestamp'])
```

```python
df['published_year'] = df['published_timestamp'].dt.year
```
