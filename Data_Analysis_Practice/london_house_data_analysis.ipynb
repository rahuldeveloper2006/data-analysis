### A. Convert the datatype of 'Date' column to date-time format.

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df.info()
```

### B. Add a new column "year" in the data frame , which contains years only.

```python
df.sample(5)
```

```python
df['year'] = df['date'].dt.year
```

```python
df.sample(3)
```

### C. add a new column "month" as 2nd column in the dataframe , which contains month only.

```python
df.insert(1 , 'month' , df['date'].dt.month)
```

```python
df.sample(4)
```

### D. Remove the column "year' and "month" from the dataframe.

```python
df.drop(['year' , 'month'] , axis = 1 , inplace=True)
```

### F. What is the maximum and minimum 'average-price' per year in england ?

```python
df1 = df[df['area'] == 'england']
df1['year'] = df1['date'].dt.year
df1
```

```python
print("maximum average price per year in england")
df1.groupby('year')['average_price'].max()
```

```python
print("minimum average price per year in england : ")
df1.groupby('year')['average_price'].min()
```

### G. What is the maximum and minimum no. of crimes recorded per area ?

```python
# now i find maximum crime records per area
print("maximum Crime records per area : ")
df.groupby('area')['no_of_crimes'].max()
```

```python
# now i find minimum crime records per area.
print("minimum Crime records per area : ")
df.groupby('area')['no_of_crimes'].min()
```
