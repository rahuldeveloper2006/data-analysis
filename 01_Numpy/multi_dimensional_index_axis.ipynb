## Broad Casting

```python
import numpy as np

matrix = np.array([[1, 2], 
                   [3, 4]])
vector = np.array([10, 20])

result = np.multiply(matrix, vector)
print(result)
# Output:
# [[10 40]    --> (1*10, 2*20)
#  [30 80]]   --> (3*10, 4*20)
```
