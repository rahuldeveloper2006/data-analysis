# Now we compare between Numpy and Python List
lets we compare by time module

```python
import numpy as np
import time

# now compare to the python list
size= 1_000_000
l1=list(range(size))
l2=list(range(size))
start = time.time()
add = [x+y for x,y in zip(l1,l2)]
end=time.time()
print("here the python list Time taken :",end-start)
```

```python
#now compare to the  Numpy array
num1=np.array(list(range(size)))
num2=np.array(list(range(size)))
start=time.time()
add=num1+num2
end=time.time()
print("here the Numpy array time taken : ",end-start)
```

```python
lists=list(range(100))
start=time.time()
sqare=[x**2 for x in (lists)]
end=time.time()
print("python list time taken :",end-start)

```

```python
nump=np.array(list(range(100)))
start=time.time()
squar=nump**2
end=time.time()
print("Numpy array time taken : ",end-start)

```
