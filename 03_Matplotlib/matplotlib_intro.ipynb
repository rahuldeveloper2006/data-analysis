```python
koholi=[0,300,400,200,500,600,1000]
dhoni=[300,200,400,500,700,200,100]
years=[1990,1991,1992,1993,1994,1995,1996]
plt.plot(koholi,years,label="Virat koholi")
plt.plot(dhoni,years,label="Mahendra singh Dhoni")
plt.legend()
plt.title("Cricket players Runs Score")
plt.xlabel("Run Score")
plt.ylabel("Years")


```

```python
koholi=[0,300,400,200,500,600,1000]
dhoni=[300,200,400,500,700,200,100]
years=[1990,1991,1992,1993,1994,1995,1996]
plt.plot(koholi,years,color="orange",linestyle='--',linewidth=3,label="Virat koholi")
plt.plot(dhoni,years,color="orange",linestyle="--",linewidth=3,label="Mahendra singh Dhoni")
plt.legend()
plt.title("Cricket players Runs Score")
plt.xlabel("Run Score")
plt.ylabel("Years")
plt.grid(True)
```

```python
import pandas as pd
df=pd.read_json('insta_data.json')

print(df.info())
```

```python
name=df['user_name'].head(4)

followers=df["followers"].head(4)

posts=df["posts"].head(4)
following=df["following"].head(4)
plt.bar(name,followers,label="followers")
plt.title("Insta Data analysis")
plt.xlabel("Followers")
plt.ylabel("Names")

```

```python
plt.bar(name,following ,color='red')
plt.title("Insta Data Analysis")
plt.xlabel("Names")
plt.ylabel("Following")
```

```python
plt.bar(name,posts,color='pink')
plt.title("Insta Data Analysis")
plt.xlabel("Names")
plt.ylabel("Following")
```
