```python
import pandas as pd
```

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries,on='employee_id',how ='inner') #only common 
```

```python
pd.merge(employees,salaries,on='employee_id',how ='outer') #all
```

```python
pd.merge(employees,salaries,on='employee_id',how ='right') # based on Salaries
```

```python
pd.merge(employees,salaries,on='employee_id',how ='left') #based on Employees 
```

# Concatenation :-

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
print(df1)
print("*"*50)
print(df2)
```

```python
pd.concat((df1 , df2)) #column wise concat
```

```python
pd.concat((df1 , df2) , axis = 1)
```

```python
df3 = pd.DataFrame({
    'A1': ['A', 'A', 'A' , 'A'],
    'B1': ['B', 'B', 'B' , 'B'],
    'C1': ['C', 'C', 'C' , 'C'],
    'D1': ['D', 'D', 'D' , 'D']
})
df3
```

```python
pd.concat((df1 , df3))
```

```python
pd.concat((df1 , df3) , axis = 1)
```
