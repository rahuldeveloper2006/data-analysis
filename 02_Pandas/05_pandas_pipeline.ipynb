# Pandas Pipe Line


Prompt:

I have a raw sales CSV file with the following columns:
Order_ID, Order_Date, Customer_Name, City, State, Product_Category, Product_Name, Quantity, Unit_Price, Discount, Payment_Mode, Sales_Channel.

Create a Pandas automation pipeline to perform the following tasks:

Clean text columns such as Customer_Name, City, and State by removing extra spaces and formatting them properly.

Remove duplicate records based on Order_ID.

Convert numeric columns (Quantity, Unit_Price, Discount) to proper numeric data types.

Create a new column Total_Sales using the formula:

Total_Sales = (Quantity * Unit_Price) - Discount.

Categorize orders into High, Medium, and Low value based on Total_Sales.

Generate summary reports:
Total sales by City
Total sales by Product_Category
Sort all summary results in descending order.
Export the cleaned dataset and summary reports into separate CSV files.

Use Pandas best practices and implement the solution using a clean and reusable pipeline approach.

```python
# its solution
import pandas as pd
#cleaning function
def clean_data(df):
    return(df
           #clean text columns 
           .assign(Customer_Name=lambda x: x["Customer_Name"].str.strip().str.title(),
                   City=lambda x: x["City"].str.strip().str.title(),
                   State=lambda x: x["State"].str.strip().str.title())
           .drop_duplicates(subset="Order_ID")
           #Convert numeric columns
           .assign(Quantity=lambda x:pd.to_numeric(x["Quantity"],errors="coerce"),
                   Unit_Price=lambda x:pd.to_numeric(x["Unit_Price"],errors="coerce"),
                   Discount=lambda x:pd.to_numeric(x["Discount"],errors="coerce"))
        #Create a new column Total_Sales using the formulla
           .assign(Total_Sales=lambda x: (x["Quantity"]*x["Unit_Price"])-x["Discount"])
           #categoroze Orders
           .assign(Category=lambda x:pd.cut(x["Total_Sales"],bins=[-float("inf"),1000,5000,float("inf")],labels=["Low","Medium","High"]))
          )
# Now we generate summery Reports
def create_summery(df):
    #Total sales by city
    city_summery=(df.groupby("City")["Total_Sales"].sum().sort_values(ascending=False).reset_index())
    product_summery=(df.groupby("Product_Category")["Total_Sales"].sum().sort_values(ascending=False).reset_index())
    return city_summery,product_summery

#Automatiom Function
def process_pipeline(file):
    df=pd.read_csv(file)
    df_cleaned=clean_data(df)
    #create summery
    city_summery,product_summery=create_summery(df_cleaned)
    df_cleaned.to_csv("cleaned_data.csv",index=False)
    city_summery.to_csv("city_summery_data.csv",index=False)
    product_summery.to_csv("product_summery_data.csv",index=False)
    return df_cleaned,city_summery,product_summery

# Run Pipe Line
df_cleaned,city_summery,product_summery=process_pipeline("raw_sales_data.csv")
    

    
                   

```
