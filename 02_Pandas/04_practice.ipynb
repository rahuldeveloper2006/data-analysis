```python
# Now we Read a Csv file 
import pandas as pd
df=pd.read_csv('day24_Employee_Data.csv')
#print(df)

#for find top rows in data frame
print(df.head())  # itshows bydefault upper 5 rows
print(df.head(15))  # it shows upper 15 Rows

# For show lower rows , then
print(df.tail())  # it shows bydefault 5 lower Rows
print(df.tail(15)) # it shows 15 lower rows
```

```python
# For find Row and Column of the Data Frame , Then
print(df.shape) # it shows number of ROWS AND COLUMNS
print(df.columns)  # here shows all columns
print(df.info())  # this method show the information of data
```

```python
# now we clean the data by removing extra spaces and write capital letter first letter of the word
print(df.head())
df['Employee_Name']=df['Employee_Name'].str.strip().str.capitalize()
print(df.head())
df['City']=df['City'].str.strip().str.title()
print(df.head())
```

```python
print(df.head(3))
dm=df.head(3).copy()
print(dm)
dm['Employee_Name']=dm['Employee_Name'].str.upper()
print(dm)
print(df.head(3))
print(dm.columns.shape)
```

```python
count_dept=df.groupby('Department')['Employee_ID'].sum()
print(count_dept)  # in this case sum the employee id by string concatenation
count_dept=df.groupby('Department')['Employee_Name'].sum()
print(count_dept) # in this case sum the all employee names by string concatenation
count_dept=df.groupby('Department')['Employee_ID'].count()
print(count_dept)   # in this case count the employees
```
