# Core Data Structure in Pandas
## Series 1D Labeled Array

```python
import pandas as pd
arr=pd.Series([12,34,56,78,23])
```

```python
print(arr)
```

```python
arr2=pd.Series([18,12,13,14,16,17],index=["Rahul","Rohon","Ritik","Roshan","Rakhi","Rakesh"])
print(arr2)
```

```python
print(arr2["Roshan"])
```
