# 📊 Seaborn – Bar Plot & Count Plot

# 1. Bar Plot

## Definition
A Bar Plot is used to show the relationship between a categorical variable and a numerical variable.

By default, it displays the **mean (average)** of the numerical variable for each category.

### Uses
- Compare average values across categories.
- Analyze sales, salary, marks, etc.
- Compare numerical data between groups.

### Basic Syntax

```python
sns.barplot(data=df, x='category', y='value')
plt.show()
```

### Important Parameters & Possible Values

| Parameter | Possible Values |
|---|---|
| data | DataFrame |
| x, y | Column names |
| hue | Categorical column |
| estimator | `np.mean`, `np.median`, `sum`, `len` |
| errorbar | `'ci'`, `'sd'`, `'se'`, `'pi'`, `None` |
| orient | `'v'`, `'h'`, `None` |
| palette | `'deep'`, `'muted'`, `'Set1'`, `'Set2'`, `'viridis'` |
| order | List of categories |
| hue_order | List of hue categories |
| dodge | `True`, `False`, `'auto'` |
| width | Positive float (default 0.8) |
| saturation | 0 to 1 |
| capsize | Float or `None` |
| fill | `True`, `False` |
| alpha | 0 to 1 |

### Practical Examples

**Basic Bar Plot**
```python
sns.barplot(data=df, x='day', y='total_bill')
```

**With Hue**
```python
sns.barplot(data=df, x='day', y='total_bill',
            hue='sex', palette='Set2')
```

**Using Median**
```python
sns.barplot(data=df, x='day', y='total_bill',
            estimator=np.median)
```

**Remove Error Bars**
```python
sns.barplot(data=df, x='day', y='total_bill',
            errorbar=None)
```

# 2. Count Plot

## Definition
A Count Plot is used to display the number of observations in each category.

It counts the frequency of each categorical value and represents it using bars.

### Uses
- Count categorical values.
- Analyze gender distribution.
- Count students in different departments.
- Understand category frequency.

### Basic Syntax

```python
sns.countplot(data=df, x='category')
plt.show()
```

### Important Parameters & Possible Values

| Parameter | Possible Values |
|---|---|
| data | DataFrame |
| x, y | Categorical column |
| hue | Another categorical column |
| orient | `'v'`, `'h'`, `None` |
| order | List of categories |
| hue_order | List of hue categories |
| palette | `'deep'`, `'muted'`, `'Set2'`, `'viridis'` |
| dodge | `True`, `False`, `'auto'` |
| stat | `'count'`, `'percent'`, `'proportion'`, `'probability'` |
| width | Positive float (default 0.8) |
| saturation | 0 to 1 |
| fill | `True`, `False` |

### Practical Examples

**Basic Count Plot**
```python
sns.countplot(data=df, x='day')
```

**With Hue**
```python
sns.countplot(data=df, x='day', hue='sex')
```

**Horizontal Count Plot**
```python
sns.countplot(data=df, y='day')
```

**Show Percentage**
```python
sns.countplot(data=df, x='day', stat='percent')
```

---

# 3. Difference Between Bar Plot and Count Plot

| Bar Plot | Count Plot |
|---|---|
| Shows numerical aggregation | Shows frequency |
| Requires categorical and numerical variables | Requires categorical variable |
| Default estimator is mean | Default statistic is count |
| Uses `sns.barplot()` | Uses `sns.countplot()` |

## 4. Quick Revision

- Bar Plot → Category vs Numerical value.
- Count Plot → Category vs Frequency.
- `estimator=np.mean` → Average value.
- `estimator=np.median` → Median value.
- `hue` → Group comparison using colors.
- `errorbar=None` → Remove error bars.
- `stat='percent'` → Show percentage instead of count.

**Functions:**
- `sns.barplot()`
- `sns.countplot()`
