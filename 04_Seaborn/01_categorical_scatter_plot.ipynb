# 📊 Seaborn – Categorical Scatter Plots

## 1. What are Categorical Scatter Plots?

Categorical Scatter Plots are used to visualize the distribution of numerical data across different categorical groups.

They help us understand how numerical values are distributed within different categories.

**Example:** Comparing the salary distribution of employees across different departments.

### Types:
1. Strip Plot
2. Swarm Plot

---

# 2. Strip Plot

## Definition
A Strip Plot is a categorical scatter plot that represents individual data points as dots along a categorical axis.

It helps visualize the distribution of numerical values across different categories.

### Syntax

```python
sns.stripplot(data=df, x='category', y='value')
plt.show()
```

### Important Parameters

| Parameter | Purpose |
|---|---|
| data | Input dataset |
| x | Categorical variable |
| y | Numerical variable |
| hue | Add another categorical variable using colors |
| jitter | Add random displacement to avoid overlapping |
| dodge | Separate hue groups |
| size | Control marker size |
| alpha | Control transparency |
| palette | Change colors |
| orient | Set plot orientation |

### Example 1: Basic Strip Plot

```python
sns.stripplot(data=df, x='day', y='total_bill')
plt.show()
```

### Example 2: Strip Plot with Hue

```python
sns.stripplot(
    data=df,
    x='day',
    y='total_bill',
    hue='sex',
    palette='Set2'
)
plt.show()
```

### Example 3: Strip Plot with Jitter

```python
sns.stripplot(
    data=df,
    x='day',
    y='total_bill',
    jitter=True,
    size=5
)
plt.show()
```

**Remember:** Strip Plot uses jitter to spread overlapping points randomly.

---

# 3. Swarm Plot

## Definition
A Swarm Plot is a categorical scatter plot that displays individual data points without overlapping.

It automatically adjusts the position of points to avoid overlapping while keeping them close to their actual categorical position.

### Syntax

```python
sns.swarmplot(data=df, x='category', y='value')
plt.show()
```

### Important Parameters

| Parameter | Purpose |
|---|---|
| data | Input dataset |
| x | Categorical variable |
| y | Numerical variable |
| hue | Differentiate groups using colors |
| dodge | Separate hue groups |
| size | Control marker size |
| palette | Change colors |
| orient | Set plot orientation |
| order | Specify category order |

### Example 1: Basic Swarm Plot

```python
sns.swarmplot(data=df, x='day', y='total_bill')
plt.show()
```

### Example 2: Swarm Plot with Hue

```python
sns.swarmplot(
    data=df,
    x='day',
    y='total_bill',
    hue='sex',
    palette='Set2'
)
plt.show()
```

### Example 3: Custom Category Order

```python
sns.swarmplot(
    data=df,
    x='day',
    y='total_bill',
    order=['Thur', 'Fri', 'Sat', 'Sun']
)
plt.show()
```

**Remember:** Swarm Plot automatically positions points to minimize overlap.

---

# 4. Difference Between Strip Plot and Swarm Plot

| Strip Plot | Swarm Plot |
|---|---|
| Points may overlap | Points are positioned to avoid overlap |
| Uses jitter for spreading points | Uses automatic point positioning |
| Faster for large datasets | Can be slower for large datasets |
| Good for general distribution | Good for detailed distribution |
| Supports many observations more easily | May become crowded with many observations |

---

# 5. When Should I Use Them?

### Use Strip Plot when:
- You want to visualize individual observations.
- You want a simple categorical scatter plot.
- Your dataset contains many observations.
- Some point overlap is acceptable.

### Use Swarm Plot when:
- You want to see every observation clearly.
- You want to minimize overlapping.
- Your dataset is relatively small or moderate.
- You want to understand the distribution within each category.

---

# 6. Important Points to Remember

1. Both plots are categorical scatter plots.
2. Both visualize numerical values against categorical groups.
3. Strip Plot uses jitter to reduce overlapping.
4. Swarm Plot automatically arranges points to avoid overlap.
5. Both support hue for additional categorical grouping.
6. Swarm Plot may become slow or crowded with large datasets.

**Quick Revision:**

- Strip Plot → Categorical distribution + Jitter.
- Swarm Plot → Categorical distribution + Non-overlapping points.

**Library:** Seaborn

**Functions:** `sns.stripplot()` and `sns.swarmplot()`
