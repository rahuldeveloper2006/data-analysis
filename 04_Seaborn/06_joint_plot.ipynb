# 🔥 SEABORN JOINTPLOT – QUICK REVISION NOTES

## 1. What is a Jointplot?

A Jointplot is a Seaborn visualization used to show the relationship between two numerical variables.

It combines:
- **Joint Distribution:** Shows the relationship between two variables in the center.
- **Marginal Distribution:** Shows the individual distribution of each variable on the top and right sides.

**Remember:** Jointplot = Scatter/Distribution Plot + Marginal Distribution Plots.

## 2. Basic Syntax

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.jointplot(data=df, x='column1', y='column2')
plt.show()
```

## 3. Important Parameters

| Parameter | Purpose |
|---|---|
| `data` | Dataset |
| `x` | X-axis numerical column |
| `y` | Y-axis numerical column |
| `kind` | Type of joint plot |
| `hue` | Differentiate categories using colors |
| `height` | Figure height |
| `ratio` | Ratio of joint and marginal plot sizes |
| `space` | Space between joint and marginal plots |
| `color` | Set plot color |
| `marginal_ticks` | Show marginal axis ticks |

## 4. Types of Jointplot (Most Important)

### A. Scatter Plot (Default)

Displays the relationship between two numerical variables.

```python
sns.jointplot(
    data=df,
    x='total_bill',
    y='tip',
    kind='scatter'
)

plt.show()
```

### B. Hexbin Plot

Displays data density using hexagonal bins.

```python
sns.jointplot(
    data=df,
    x='total_bill',
    y='tip',
    kind='hex'
)

plt.show()
```

### C. KDE Plot

Displays the density distribution of two numerical variables.

```python
sns.jointplot(
    data=df,
    x='total_bill',
    y='tip',
    kind='kde'
)

plt.show()
```

### D. Histogram Plot

Displays the distribution using histograms.

```python
sns.jointplot(
    data=df,
    x='total_bill',
    y='tip',
    kind='hist'
)

plt.show()
```

### E. Regression Plot

Displays the relationship between two variables with a regression line.

```python
sns.jointplot(
    data=df,
    x='total_bill',
    y='tip',
    kind='reg'
)

plt.show()
```

### F. Residual Plot

Displays the residuals of a linear regression model.

```python
sns.jointplot(
    data=df,
    x='total_bill',
    y='tip',
    kind='resid'
)

plt.show()
```

## 5. Using Hue Parameter

Hue helps differentiate data based on categorical values.

```python
sns.jointplot(
    data=df,
    x='total_bill',
    y='tip',
    hue='sex'
)

plt.show()
```

Note: Hue support depends on the selected plot kind and Seaborn version.

## 6. Change Figure Size and Color

```python
sns.jointplot(
    data=df,
    x='total_bill',
    y='tip',
    kind='scatter',
    height=8,
    ratio=5,
    space=0.2,
    color='blue'
)

plt.show()
```

## 7. When Should We Use Jointplot?

1. To analyze the relationship between two numerical variables.
2. To understand the distribution of individual variables.
3. To identify positive or negative relationships.
4. To identify data density and concentration.
5. To observe regression relationships.
6. To detect possible outliers.

## 8. Important Things to Remember

- Jointplot works mainly with two numerical variables.
- `x` and `y` define the two variables.
- `kind` controls the central plot type.
- Scatter is the default kind.
- Marginal plots show individual distributions.
- KDE represents probability density.
- Hexbin is useful for large datasets with overlapping points.
- Regression kind displays a fitted regression line.
- Jointplot returns a JointGrid object.

## 9. Quick Revision Cheat Sheet

```python
# Basic jointplot
sns.jointplot(data=df, x='col1', y='col2')

# Scatter jointplot
sns.jointplot(data=df, x='col1', y='col2', kind='scatter')

# Hexbin jointplot
sns.jointplot(data=df, x='col1', y='col2', kind='hex')

# KDE jointplot
sns.jointplot(data=df, x='col1', y='col2', kind='kde')

# Histogram jointplot
sns.jointplot(data=df, x='col1', y='col2', kind='hist')

# Regression jointplot
sns.jointplot(data=df, x='col1', y='col2', kind='reg')

# Residual jointplot
sns.jointplot(data=df, x='col1', y='col2', kind='resid')

# Jointplot with hue
sns.jointplot(data=df, x='col1', y='col2', hue='category')
```

**REMEMBER:** Jointplot = Relationship between two variables + Individual distributions.
