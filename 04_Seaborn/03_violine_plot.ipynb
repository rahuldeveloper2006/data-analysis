# 🎻 Seaborn – Violin Plot

## 1. Definition
A Violin Plot is a categorical visualization that combines a Box Plot and KDE (Kernel Density Estimation).

It shows the distribution and density of numerical data across different categories.

**Uses:**
- Understand data distribution.
- Compare numerical values across categories.
- Identify data concentration and multiple peaks.

## 2. Basic Syntax

```python
sns.violinplot(data=df, x='category', y='value')
plt.show()
```

## 3. Important Parameters & Possible Values

| Parameter | Possible Values / Meaning |
|---|---|
| data | DataFrame |
| x, y | Column names |
| hue | Categorical column |
| orient | `'v'`, `'h'`, `None` |
| palette | `'deep'`, `'muted'`, `'Set1'`, `'Set2'`, `'viridis'` |
| inner | `'box'`, `'quart'`, `'point'`, `'stick'`, `None` |
| split | `True`, `False` |
| width | Positive float (default 0.8) |
| cut | Non-negative float (default 2) |
| bw_method | `'scott'`, `'silverman'`, float, callable |
| bw_adjust | Positive float (default 1) |
| density_norm | `'area'`, `'count'`, `'width'` |
| dodge | `True`, `False`, `'auto'` |
| fill | `True`, `False` |
| linewidth | Positive float |
| saturation | 0 to 1 |
| order | List of category names |
| hue_order | List of hue categories |

## 4. Practical Examples

**Basic Violin Plot**
```python
sns.violinplot(data=df, x='day', y='total_bill')
```

**With Hue**
```python
sns.violinplot(data=df, x='day', y='total_bill',
               hue='sex', palette='Set2')
```

**Split Violin Plot**
```python
sns.violinplot(data=df, x='day', y='total_bill',
               hue='sex', split=True)
```

**Change Inner Style**
```python
sns.violinplot(data=df, x='day', y='total_bill',
               inner='quart')
```

**Control KDE Smoothing**
```python
sns.violinplot(data=df, x='day', y='total_bill',
               bw_adjust=2, cut=0)
```

## 5. Important Concepts

- Wider region = Higher data density.
- Narrower region = Lower data density.
- `inner='box'` = Shows box plot inside violin.
- `inner='quart'` = Shows quartile lines.
- `inner='point'` = Shows individual observations.
- `split=True` = Displays hue groups as violin halves.
- `bw_adjust` = Controls KDE smoothness.
- `cut=0` = Prevents KDE from extending beyond observed data.

## 6. Box Plot vs Violin Plot

| Box Plot | Violin Plot |
|---|---|
| Shows median and quartiles | Shows data density |
| Statistical summary | Distribution shape |
| Does not show KDE | Uses KDE |

**Remember:** Violin Plot = Box Plot + KDE.

**Function:** `sns.violinplot()`

````python
sns.violinplot(data = df , x= 'day' , y = 'total_bill' , hue = 'sex' , split=True)
````
