# 📦 Seaborn – Box Plot

## 1. What is a Box Plot?

A Box Plot is a statistical visualization used to understand the distribution of numerical data through five important values:

1. Minimum (Lower whisker)
2. First Quartile (Q1 – 25th percentile)
3. Median (Q2 – 50th percentile)
4. Third Quartile (Q3 – 75th percentile)
5. Maximum (Upper whisker)

It is mainly used to identify outliers, understand data spread, and compare distributions across different categories.

**Main Purpose:**
- Detect outliers.
- Understand data distribution.
- Identify median and quartiles.
- Compare numerical data across categories.
- Understand the spread of data.

---

## 2. Understanding Box Plot Components

### Five Number Summary

| Component | Meaning |
|---|---|
| Minimum | Lowest non-outlier value |
| Q1 | 25th percentile |
| Median (Q2) | 50th percentile |
| Q3 | 75th percentile |
| Maximum | Highest non-outlier value |

### Important Formulas

**Interquartile Range (IQR):**

IQR = Q3 - Q1

**Lower Bound:**

Lower Bound = Q1 - 1.5 × IQR

**Upper Bound:**

Upper Bound = Q3 + 1.5 × IQR

Values below the lower bound or above the upper bound are considered potential outliers.

Note: Whiskers extend to the most extreme observations within these limits, not necessarily the actual minimum and maximum.

---

## 3. Basic Syntax

```python
sns.boxplot(data=df, x='category', y='value')
plt.show()
```

You can also use a single numerical column:

```python
sns.boxplot(data=df, y='total_bill')
plt.show()
```

---

## 4. Important Parameters and Their Possible Values

### 1. data
Specifies the input dataset.

**Possible values:**
- pandas DataFrame
- pandas Series
- Long-form dataset

```python
sns.boxplot(data=df)
```

### 2. x
Specifies the variable displayed on the X-axis.

**Possible values:**
- Column name (string)
- Numerical column
- Categorical column

```python
sns.boxplot(data=df, x='day')
```

### 3. y
Specifies the variable displayed on the Y-axis.

**Possible values:**
- Column name (string)
- Numerical column
- Categorical column (depending on orientation)

```python
sns.boxplot(data=df, y='total_bill')
```

### 4. hue
Used to divide data into additional groups using different colors.

**Possible values:**
- Any categorical column name.
- None (default).

```python
sns.boxplot(data=df, x='day', y='total_bill', hue='sex')
```

### 5. orient
Controls the orientation of the box plot.

**Possible values:**
- `'v'` → Vertical
- `'h'` → Horizontal
- `None` → Automatically determine orientation.

```python
sns.boxplot(data=df, y='total_bill', orient='v')
```

### 6. order
Controls the order of categorical values.

**Possible values:**
- List of category names.
- None (default).

```python
sns.boxplot(
    data=df,
    x='day',
    y='total_bill',
    order=['Thur', 'Fri', 'Sat', 'Sun']
)
```

### 7. hue_order
Controls the order of categories in the hue variable.

**Possible values:**
- List of category names.
- None (default).

```python
sns.boxplot(
    data=df,
    x='day',
    y='total_bill',
    hue='sex',
    hue_order=['Female', 'Male']
)
```

### 8. palette
Controls the colors used for different categories.

**Possible values:**

Named palettes:
- `'deep'`
- `'muted'`
- `'bright'`
- `'pastel'`
- `'dark'`
- `'colorblind'`

Other options:
- `'Set1'`
- `'Set2'`
- `'Set3'`
- `'viridis'`
- `'coolwarm'`

You can also provide a list or dictionary of colors.

```python
sns.boxplot(
    data=df,
    x='day',
    y='total_bill',
    palette='Set2'
)
```

### 9. saturation
Controls color saturation.

**Possible values:**
- Float between `0` and `1`.
- Default: `0.75`.

```python
sns.boxplot(data=df, x='day', y='total_bill', saturation=0.5)
```

### 10. width
Controls the width of the boxes.

**Possible values:**
- Positive float.
- Default: `0.8`.

```python
sns.boxplot(data=df, x='day', y='total_bill', width=0.5)
```

### 11. dodge
Controls whether boxes for different hue groups are separated.

**Possible values:**
- `True` → Separate hue groups.
- `False` → Do not separate hue groups.
- `'auto'` → Automatically determine separation in supported versions.

```python
sns.boxplot(
    data=df,
    x='day',
    y='total_bill',
    hue='sex',
    dodge=True
)
```

### 12. whis
Controls the whisker length using the IQR.

**Possible values:**
- Float: `1.5` (default).
- Any non-negative float, e.g. `1`, `2`, `3`.
- Tuple of percentiles, e.g. `(5, 95)`.
- `'range'` → Extend whiskers to minimum and maximum.

```python
sns.boxplot(data=df, y='total_bill', whis=1.5)
```

**Examples:**

- `whis=1.5` → Standard Tukey whiskers.
- `whis=2` → Whiskers extend further.
- `whis=(10, 90)` → Whiskers use 10th and 90th percentiles.
- `whis='range'` → Whiskers cover the full data range.

### 13. fliersize
Controls the size of outlier markers.

**Possible values:**
- Non-negative float.
- Default: `5`.

```python
sns.boxplot(data=df, y='total_bill', fliersize=3)
```

### 14. linewidth
Controls the thickness of the box plot lines.

**Possible values:**
- Positive float.
- Default: `None` (uses plotting defaults).

```python
sns.boxplot(data=df, y='total_bill', linewidth=2)
```

### 15. showfliers
Controls whether outlier points are displayed.

**Possible values:**
- `True` → Show outliers.
- `False` → Hide outliers.

```python
sns.boxplot(data=df, y='total_bill', showfliers=False)
```

### 16. fill
Controls whether the boxes are filled with color (available in newer Seaborn versions).

**Possible values:**
- `True` → Filled boxes.
- `False` → Unfilled boxes.

```python
sns.boxplot(data=df, x='day', y='total_bill', fill=False)
```

## 6. When Should We Use Box Plot?

Use Box Plot when:

1. You want to detect potential outliers.
2. You want to understand the spread of numerical data.
3. You want to compare salary distributions across departments.
4. You want to compare marks across different classes.
5. You want to analyze the distribution of numerical variables.

## 7. Important Limitations

- It does not show every individual observation.
- Different distributions can have similar box plots.
- It does not directly show the frequency distribution.
- Outliers are potential unusual observations, not necessarily incorrect data.
