# 🔥 SEABORN HEATMAP – QUICK REVISION NOTES

## 1. What is a Heatmap?

A Heatmap is a data visualization technique used to represent numerical values using different colors.

- Different colors represent different numerical values.
- It helps identify patterns, relationships, and correlations in a dataset.
- Mostly used for correlation analysis.

## 2. Basic Syntax

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(data)
plt.show()
```

## 3. Important Parameters

| Parameter | Purpose |
|---|---|
| `data` | 2D numerical data |
| `annot=True` | Display values inside cells |
| `fmt='.2f'` | Format values to 2 decimal places |
| `cmap='coolwarm'` | Set color theme |
| `linewidths=0.5` | Add space between cells |
| `linecolor='black'` | Set cell border color |
| `vmin=-1` | Minimum color scale value |
| `vmax=1` | Maximum color scale value |
| `cbar=True` | Display color bar |
| `square=True` | Make cells square |
| `mask=` | Hide selected cells |

## 4. Most Important: Correlation Heatmap

Correlation measures the relationship between two numerical variables.

**Correlation values:**

- `+1` → Perfect Positive Correlation
- `0` → No Linear Correlation
- `-1` → Perfect Negative Correlation

### Example:

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(10, 6))

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5,
    linecolor='black',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap')
plt.show()
```

## 5. Hide Upper Triangle (Advanced)

Useful when you want to avoid displaying duplicate correlation values.

```python
import numpy as np

corr = df.corr(numeric_only=True)

mask = np.triu(np.ones_like(corr, dtype=bool))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    mask=mask
)

plt.show()
```

## 6. Change Heatmap Color

```python
sns.heatmap(df.corr(numeric_only=True), cmap='viridis')
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm')
sns.heatmap(df.corr(numeric_only=True), cmap='YlGnBu')
sns.heatmap(df.corr(numeric_only=True), cmap='magma')
```

## 7. When Should We Use Heatmap?

1. To find correlation between numerical columns.
2. To identify highly correlated features.
3. To visualize missing values.
4. To identify patterns in a matrix.
5. To analyze relationships in a dataset.

### Missing Values Heatmap

```python
sns.heatmap(df.isnull(), cmap='viridis', cbar=False)
plt.show()
```

## 8. Important Things to Remember

- Heatmap generally requires 2D data.
- Correlation heatmap works with numerical columns.
- `annot=True` displays values.
- `fmt='.2f'` controls decimal formatting.
- `cmap` controls colors.
- `df.corr()` calculates correlation.
- Correlation does NOT mean causation.

## 9. Quick Revision Cheat Sheet

```python
# Correlation matrix

# Basic heatmap

# Heatmap with values

# Heatmap with borders

# Missing values visualization

```

**REMEMBER:** Heatmap = Numerical Data + Color Representation + Pattern Identification.

# 🔥 SEABORN CLUSTERMAP – QUICK REVISION NOTES

## 1. What is a Clustermap?

A Clustermap is a data visualization technique that combines a heatmap with hierarchical clustering.

- It groups similar rows and columns together.
- It identifies patterns and similarities in data.
- It rearranges rows and columns based on their similarity.
- It uses a dendrogram to show hierarchical relationships.

**Remember:** Heatmap displays values, whereas Clustermap displays values + similarity-based grouping.

## 2. Basic Syntax

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.clustermap(data)
plt.show()
```

## 3. Important Parameters

| Parameter | Purpose |
|---|---|
| `data` | 2D numerical data |
| `method` | Linkage method |
| `metric` | Distance measurement |
| `cmap` | Color theme |
| `annot` | Display values (not supported directly by clustermap) |
| `figsize` | Figure dimensions |
| `standard_scale` | Scale data between 0 and 1 |
| `z_score` | Apply Z-score normalization |
| `row_cluster` | Enable/disable row clustering |
| `col_cluster` | Enable/disable column clustering |
| `linewidths` | Width of cell borders |
| `cbar_pos` | Position of color bar |

## 4. Basic Clustermap Example

```python
sns.clustermap(
    df.corr(numeric_only=True),
    cmap='coolwarm',
    figsize=(10, 8)
)

plt.show()
```

## 5. Clustermap with Different Linkage Methods

Linkage methods determine how clusters are formed.

```python
sns.clustermap(data, method='average')
sns.clustermap(data, method='single')
sns.clustermap(data, method='complete')
sns.clustermap(data, method='ward')
```

Common methods:
- `single` – Minimum distance between clusters.
- `complete` – Maximum distance between clusters.
- `average` – Average distance between clusters.
- `ward` – Minimizes within-cluster variance (typically with Euclidean distance).

## 6. Distance Metrics

Distance metrics measure the difference between observations.

```python
sns.clustermap(data, metric='euclidean')
sns.clustermap(data, metric='cityblock')
sns.clustermap(data, metric='correlation')
```

Common metrics:
- Euclidean – Straight-line distance.
- Cityblock – Manhattan distance.
- Correlation – Distance based on correlation.

## 7. Standardization

Useful when columns have different scales.

### Z-score normalization

```python
sns.clustermap(
    data,
    z_score=1,
    cmap='coolwarm'
)

plt.show()
```

`z_score=1` standardizes columns.

`z_score=0` standardizes rows.

### Min-Max scaling

```python
sns.clustermap(
    data,
    standard_scale=1,
    cmap='viridis'
)

plt.show()
```

`standard_scale=1` scales columns between 0 and 1.

`standard_scale=0` scales rows between 0 and 1.

## 8. Disable Row or Column Clustering

```python
# Only column clustering
sns.clustermap(data, row_cluster=False)

# Only row clustering
sns.clustermap(data, col_cluster=False)

# Disable both
sns.clustermap(
    data,
    row_cluster=False,
    col_cluster=False
)
```

## 9. Important Differences: Heatmap vs Clustermap

| Heatmap | Clustermap |
|---|---|
| Displays values using colors | Displays values using colors |
| Does not automatically group similar rows | Groups similar rows |
| No dendrogram | Includes dendrogram |
| Simple visualization | Hierarchical clustering visualization |
| Uses sns.heatmap() | Uses sns.clustermap() |

## 10. Important Things to Remember

- Clustermap performs hierarchical clustering.
- It requires a 2D numerical dataset.
- Dendrogram represents hierarchical relationships.
- `method` controls the linkage technique.
- `metric` controls distance calculation.
- Standardization is useful when features have different scales.
- Ward linkage is generally used with Euclidean distance.
- Clustermap returns a ClusterGrid object.

## 11. Quick Revision Cheat Sheet

```python
# Basic clustermap
sns.clustermap(data)

# Correlation clustermap
sns.clustermap(df.corr(numeric_only=True))

# Change color
sns.clustermap(data, cmap='coolwarm')

# Change linkage
sns.clustermap(data, method='average')

# Change distance
sns.clustermap(data, metric='euclidean')

# Standardize columns
sns.clustermap(data, z_score=1)

# Scale columns between 0 and 1
sns.clustermap(data, standard_scale=1)

# Disable row clustering
sns.clustermap(data, row_cluster=False)

# Disable column clustering
sns.clustermap(data, col_cluster=False)
```

**REMEMBER:** Clustermap = Heatmap + Hierarchical Clustering + Dendrogram.
