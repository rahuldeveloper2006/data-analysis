# 🔥 SEABORN PAIRPLOT – QUICK REVISION NOTES

## 1. What is Pairplot?

Pairplot is a Seaborn visualization used to analyze relationships between multiple numerical variables in a dataset.

- It creates pairwise relationships between numerical columns.
- Diagonal plots show individual distributions.
- Off-diagonal plots show relationships between two variables.
- Useful for Exploratory Data Analysis (EDA).

**Remember:** Pairplot = Multiple Scatterplots + Distribution Plots.

## 2. Basic Syntax

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(df)
plt.show()
```

## 3. Important Parameters

| Parameter | Purpose |
|---|---|
| `data` | Dataset |
| `hue` | Differentiate categories using colors |
| `vars` | Select specific columns |
| `x_vars` | Select X-axis variables |
| `y_vars` | Select Y-axis variables |
| `kind` | Type of relationship plot |
| `diag_kind` | Type of diagonal distribution plot |
| `corner=True` | Display only lower triangle |
| `palette` | Change color palette |
| `height` | Size of each subplot |

## 4. Important Examples

**A. Pairplot with Hue**

```python
sns.pairplot(df, hue='species')
plt.show()
```

**B. Select Specific Columns**

```python
sns.pairplot(
    df,
    vars=['sepal_length', 'sepal_width', 'petal_length']
)
```

**C. Change Plot Type**

```python
sns.pairplot(
    df,
    kind='reg',
    diag_kind='kde'
)
```

**D. Display Only Lower Triangle**

```python
sns.pairplot(df, corner=True)
```

## 5. Important Things to Remember

- Pairplot is mainly used for numerical columns.
- `hue` helps compare different categories.
- `kind='scatter'` is the default.
- `kind='reg'` adds regression lines.
- `diag_kind='hist'` displays histograms on the diagonal.
- `diag_kind='kde'` displays density curves.
- Pairplot returns a PairGrid object.
- Pairplot can become slow with many numerical columns.

## 6. Quick Revision Cheat Sheet

```python
# Basic pairplot
sns.pairplot(df)

# With categorical hue
sns.pairplot(df, hue='species')

# Specific columns
sns.pairplot(df, vars=['A', 'B', 'C'])

# Regression
sns.pairplot(df, kind='reg')

# KDE on diagonal
sns.pairplot(df, diag_kind='kde')

# Lower triangle only
sns.pairplot(df, corner=True)
```

**REMEMBER:** Pairplot = Pairwise Relationships + Individual Distributions.

# 🔥 SEABORN DISTPLOT – QUICK NOTES

## 1. What is Distplot?

Distplot (Distribution Plot) is used to visualize the distribution of a numerical variable. It helps understand frequency, density, and data spread.

**Note:** `sns.distplot()` is deprecated. Use `sns.histplot()` or `sns.displot()` instead.

## 2. Basic Syntax

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(data=df, x='age', kde=True)
plt.show()
```

## 3. Important Parameters

| Parameter | Purpose |
|---|---|
| `data` | Dataset |
| `x` | Numerical column |
| `bins` | Number of histogram intervals |
| `kde=True` | Display density curve |
| `color` | Change plot color |
| `stat` | Choose count, frequency, density, etc. |
| `element` | Histogram appearance |

## 4. Important Examples

```python
# Basic histogram
sns.histplot(df['age'])

# Histogram with KDE
sns.histplot(df['age'], kde=True)

# Change bins
sns.histplot(df['age'], bins=20)

# KDE plot only
sns.kdeplot(data=df, x='age')

# Distribution plot using displot
sns.displot(data=df, x='age', kde=True)
```

## 5. Important Things to Remember

- Histogram shows frequency distribution.
- KDE shows a smoothed density curve.
- `bins` controls histogram intervals.
- Useful for identifying skewness, spread and distribution patterns.
- Distplot is deprecated in modern Seaborn.

**REMEMBER:** Distribution Plot = Histogram + KDE (optional).

````python
import warnings
warnings.filterwarnings('ignore')
data = sns.load_dataset('titanic')
data.head()
````

````python
data['age'].isna().sum()
````

````python
data['age'] = data['age'].fillna(data['age'].mean())
````

````python
sns.distplot(data['age'])
````

````python
sns.distplot(data['age'] , kde = False , bins = 20 , color='red')
````

````python
sns.distplot(data['age'] , color='red' , hist=False)
````

````python
sns.displot(data['age'] , kde = True)
````

# 🔥 SEABORN RUGPLOT – QUICK NOTES

## 1. What is Rugplot?

Rugplot is a Seaborn visualization used to display individual data points as small vertical or horizontal lines along an axis.

- Each small line represents one observation.
- Helps understand data distribution.
- Often used with KDE or histogram plots.

## 2. Basic Syntax

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.rugplot(data=df, x='age')
plt.show()
```

## 3. Important Parameters

| Parameter | Purpose |
|---|---|
| `data` | Dataset |
| `x` | Variable on X-axis |
| `y` | Variable on Y-axis |
| `height` | Length of rug lines |
| `color` | Change line color |
| `alpha` | Control transparency |
| `linewidth` | Control line thickness |

## 4. Important Examples

```python
# Basic rugplot
sns.rugplot(data=df, x='age')

# Change color
sns.rugplot(data=df, x='age', color='red')

# Change line height
sns.rugplot(data=df, x='age', height=0.1)

# Change transparency
sns.rugplot(data=df, x='age', alpha=0.5)

plt.show()
```

## 5. Combine Rugplot with KDE

```python
sns.kdeplot(data=df, x='age')
sns.rugplot(data=df, x='age')

plt.show()
```

## 6. Important Things to Remember

- Each line represents an individual observation.
- Rugplot helps identify data concentration.
- Useful for understanding distribution.
- Can be combined with histogram and KDE.
- Too many observations can make the plot crowded.

**REMEMBER:** Rugplot = Small marks representing individual data points.

````python
sns.distplot(data['age'] , kde = False , hist=False , rug=True)
````

````python
sns.rugplot(data['age'] , height=0.5 , alpha = 1)
````
