# 📊 Seaborn – Scatter Plot

## 1. What is a Scatter Plot?
A Scatter Plot is a data visualization technique used to show the relationship between two numerical variables using dots (points).

Each dot represents one observation in a dataset.

**Example:** Relationship between Study Hours and Exam Marks.

## 2. Why do we use Scatter Plot?
- To identify relationships between two numerical variables.
- To understand positive and negative correlations.
- To detect outliers.
- To identify patterns and clusters.
- To observe whether two variables are related or not.

## 3. Basic Syntax

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=df, x='column1', y='column2')
plt.show()
```

## 4. Important Parameters

| Parameter | Purpose |
|---|---|
| data | Dataset |
| x | X-axis variable |
| y | Y-axis variable |
| hue | Differentiate points using colors |
| style | Differentiate points using marker styles |
| size | Represent values using different point sizes |
| palette | Change color palette |
| alpha | Control transparency |
| legend | Show or hide legend |

## 5. Understanding Correlation Through Scatter Plot

**Positive Correlation:** Both variables increase together.

**Negative Correlation:** One variable increases while the other decreases.

**No Correlation:** No clear relationship between the variables.

## 6. Important Code Examples

### Basic Scatter Plot
```python
sns.scatterplot(data=df, x='age', y='salary')
plt.show()
```

### Scatter Plot with Hue
```python
sns.scatterplot(
    data=df,
    x='age',
    y='salary',
    hue='gender'
)
plt.show()
```

### Scatter Plot with Size
```python
sns.scatterplot(
    data=df,
    x='age',
    y='salary',
    size='experience'
)
plt.show()
```

### Scatter Plot with Style
```python
sns.scatterplot(
    data=df,
    x='age',
    y='salary',
    style='gender'
)
plt.show()
```

## 7. Important Points to Remember

- Scatter Plot is mainly used for numerical vs numerical analysis.
- `hue` adds color-based categorical differentiation.
- `size` represents another variable through marker size.
- `style` changes marker shapes.
- Scatter Plot helps identify outliers and relationships.
- Correlation does not necessarily mean causation.

## 8. When Should I Use Scatter Plot?

Use Scatter Plot when you want to investigate questions like:

1. Does study time affect exam marks?
2. Is there a relationship between age and salary?
3. Does advertising expenditure relate to sales?
4. Is there a relationship between height and weight?

**Remember:** Scatter Plot = Relationship between two numerical variables.

**Library:** Seaborn (`sns.scatterplot()`)

### Using Palette :-  palette values are : Set1 , Set2 , Set3 , viridis , rainbow

````python
sns.relplot(x= 'total_bill' , y = 'tip'  , hue = 'smoker',style='sex' , size ='size' ,palette= 'flare',sizes=(10 , 200) ,data= df , legend=False)
````

- Sequential palettes. Use rocket, mako, flare, crest, viridis, or Blues when mapping ordered numeric data from low to high.
- Diverging palettes. Use vlag, icefire, coolwarm, or RdBu when your data has a meaningful central midpoint such as zero.
- Modifiers. Append _r to any palette name (such as viridis_r) to reverse its color order, or _d on select palettes for a darker variant.
